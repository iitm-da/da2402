# Part 1: HTML Basics

## What is HTML?

**HTML** (HyperText Markup Language) is the standard language for creating web pages. Think of it as the **skeleton** of a webpage.

### HTML Elements (Tags)

HTML uses **tags** to define different parts of a page. Tags are written in angle brackets like this:

```html
<tagname>Content goes here</tagname>
```

Most tags come in pairs:
- **Opening tag**: `<tagname>`
- **Closing tag**: `</tagname>`

### Common HTML Tags

| Tag | Purpose | Example |
|-----|---------|--------|
| `<h1>` to `<h6>` | Headings (h1 is largest) | `<h1>Main Title</h1>` |
| `<p>` | Paragraph | `<p>This is a paragraph.</p>` |
| `<div>` | Container/Division | `<div>Group of elements</div>` |
| `<span>` | Inline container | `<span>Inline text</span>` |
| `<img>` | Image | `<img src="photo.jpg">` |
| `<a>` | Link | `<a href="url">Click here</a>` |
| `<ul>` / `<li>` | Unordered list | `<ul><li>Item 1</li></ul>` |
| `<table>` | Table | `<table>...</table>` |

## Example 1: Simple HTML Page

````python
from IPython.display import HTML

# Simple HTML example
html_code = """
<!DOCTYPE html>
<html>
<head>
    <title>My First Page</title>
</head>
<body>
    <h1>Welcome to HTML!</h1>
    <p>This is my first paragraph.</p>
    <p>HTML is <strong>easy</strong> to learn!</p>

    <h2>My Favorite Things:</h2>
    <ul>
        <li>Learning Python</li>
        <li>Creating dashboards</li>
        <li>Data visualization</li>
    </ul>
</body>
</html>
"""

HTML(html_code)
````

### 🔍 Understanding the Code:

- `<!DOCTYPE html>` - Tells the browser this is an HTML5 document
- `<html>` - Root element that contains all other elements
- `<head>` - Contains metadata (not visible on page)
- `<title>` - Page title (shows in browser tab)
- `<body>` - Contains all visible content
- `<h1>`, `<h2>` - Headings of different sizes
- `<p>` - Paragraphs
- `<strong>` - Bold text
- `<ul>` and `<li>` - Unordered (bulleted) list

# Part 2: CSS Basics - Styling Your Page

## What is CSS?

**CSS** (Cascading Style Sheets) controls how HTML elements look. Think of it as the **clothing and makeup** for your webpage.

### CSS Syntax

```css
selector {
    property: value;
}
```

### Common CSS Properties

| Property | Purpose | Example |
|----------|---------|--------|
| `color` | Text color | `color: blue;` |
| `background-color` | Background color | `background-color: yellow;` |
| `font-size` | Text size | `font-size: 20px;` |
| `font-weight` | Text thickness | `font-weight: bold;` |
| `padding` | Space inside element | `padding: 10px;` |
| `margin` | Space outside element | `margin: 20px;` |
| `border` | Border around element | `border: 1px solid black;` |
| `width` | Element width | `width: 300px;` |
| `height` | Element height | `height: 200px;` |

## Example 2: HTML with CSS Styling

````python
html_with_css = """
<style>
    .my-box {
        background-color: #e3f2fd;
        border: 20px solid #2196f3;
        padding: 20px;
        margin: 10px;
        border-radius: 20px;
    }

    .my-title {
        color: #1976d2;
        font-size: 24px;
        font-weight: bold;
    }

    .my-text {
        color: #424242;
        font-size: 16px;
        line-height: 1.6;
    }
</style>

<div class="my-box">
    <h2 class="my-title">Styled Box</h2>
    <p class="my-text">This box has a blue background, border, and rounded corners!</p>
    <p class="my-text">CSS makes our page look professional and organized.</p>
</div>

<div class="my-box">
    <h2 class="my-title">Another Box</h2>
    <p class="my-text">We can reuse the same styles for multiple elements.</p>
</div>
"""

HTML(html_with_css)
````

### 🔍 Understanding CSS:

- **`.my-box`** - A CSS class (can be reused on multiple elements)
- **`class="my-box"`** - Applies the CSS class to an HTML element
- **`padding`** - Space inside the box (between content and border)
- **`margin`** - Space outside the box (between boxes)
- **`border-radius`** - Makes corners rounded
- **Color codes** - `#e3f2fd` is a light blue color in hexadecimal format

# Part 3: Creating Layouts with Divs

## The `<div>` Element

The `<div>` (division) tag is the most important element for creating layouts. It's like a **container** that holds other elements.

### Think of it like boxes:
- You can put boxes inside boxes
- You can arrange boxes side-by-side or stacked
- You can style each box differently

## Example 3: Simple Grid Layout (Without Bootstrap)

````python
simple_grid = """
<style>
    .container {
        width: 100%;
        padding: 20px;
        background-color: #f5f5f5;
    }

    .row {
        display: flex;
        margin-bottom: 20px;
    }

    .box {
        flex: 1;
        padding: 20px;
        margin: 5px;
        background-color: white;
        border: 2px solid #ddd;
        text-align: center;
    }

    .box-blue { background-color: #bbdefb; }
    .box-green { background-color: #c8e6c9; }
    .box-yellow { background-color: #fff9c4; }
    .box-pink { background-color: #f8bbd0; }
</style>

<div class="container">
    <h2>Simple Grid Layout</h2>

    <!-- Row 1: Single box -->
    <div class="row">
        <div class="box box-blue">
            <h3>Full Width Box</h3>
            <p>This box takes up the entire width</p>
        </div>
    </div>

    <!-- Row 2: Two boxes side by side -->
    <div class="row">
        <div class="box box-green">
            <h3>Box 1</h3>
            <p>50% width</p>
        </div>
        <div class="box box-green">
            <h3>Box 2</h3>
            <p>50% width</p>
        </div>
    </div>

    <!-- Row 3: Three boxes -->
    <div class="row">
        <div class="box box-yellow">
            <h3>Box 1</h3>
            <p>33% width</p>
        </div>
        <div class="box box-yellow">
            <h3>Box 2</h3>
            <p>33% width</p>
        </div>
        <div class="box box-yellow">
            <h3>Box 3</h3>
            <p>33% width</p>
        </div>
    </div>

    <!-- Row 4: Four boxes -->
    <div class="row">
        <div class="box box-pink">
            <h3>1</h3>
        </div>
        <div class="box box-pink">
            <h3>2</h3>
        </div>
        <div class="box box-pink">
            <h3>3</h3>
        </div>
        <div class="box box-pink">
            <h3>4</h3>
        </div>
    </div>
</div>
"""

HTML(simple_grid)
````

### 🔍 Understanding the Layout:

- **Container** - Holds everything together
- **Row** - A horizontal line of boxes
- **Box** - Individual content areas
- **`display: flex`** - Makes boxes sit side-by-side
- **`flex: 1`** - Makes boxes share space equally

### Key Concept:
Think of a layout as a **grid of rows and columns**, like a spreadsheet!

# Part 4: Introduction to Bootstrap

## What is Bootstrap?

**Bootstrap** is a popular CSS framework that makes it easy to create professional, responsive layouts.

### Why Use Bootstrap?
- ✅ Pre-built styles and components
- ✅ Responsive (works on mobile, tablet, desktop)
- ✅ Easy grid system
- ✅ Professional look with minimal effort

## Bootstrap Grid System

Bootstrap divides the page into **12 columns**. You can combine columns to create different layouts.

### Basic Structure:
```html
<div class="container">
    <div class="row">
        <div class="col-6">Half width (6 out of 12 columns)</div>
        <div class="col-6">Half width (6 out of 12 columns)</div>
    </div>
</div>
```

### Column Sizes:
- `col-12` = Full width (12/12)
- `col-6` = Half width (6/12)
- `col-4` = One-third width (4/12)
- `col-3` = One-quarter width (3/12)

## Example 4: Bootstrap Grid Layout

````python
bootstrap_grid = """
<!-- Include Bootstrap CSS -->
<link href="https://cdn.jsdelivr.net/npm/bootstrap@5.1.3/dist/css/bootstrap.min.css" rel="stylesheet">

<style>
    .demo-box {
        background-color: #e3f2fd;
        border: 2px solid #2196f3;
        padding: 20px;
        margin-bottom: 10px;
        text-align: center;
        border-radius: 5px;
    }
</style>

<div class="container">
    <h2 class="mb-4">Bootstrap Grid System</h2>

    <!-- Full width (12 columns) -->
    <div class="row mb-3">
        <div class="col-12">
            <div class="demo-box">
                <strong>col-12</strong><br>Full Width (12 out of 12 columns)
            </div>
        </div>
    </div>

    <!-- Two equal columns (6 + 6 = 12) -->
    <div class="row mb-3">
        <div class="col-6">
            <div class="demo-box" style="background-color: #c8e6c9; border-color: #4caf50;">
                <strong>col-6</strong><br>Half Width (6/12)
            </div>
        </div>
        <div class="col-6">
            <div class="demo-box" style="background-color: #c8e6c9; border-color: #4caf50;">
                <strong>col-6</strong><br>Half Width (6/12)
            </div>
        </div>
    </div>

    <!-- Three equal columns (4 + 4 + 4 = 12) -->
    <div class="row mb-3">
        <div class="col-4">
            <div class="demo-box" style="background-color: #fff9c4; border-color: #ffc107;">
                <strong>col-4</strong><br>One-third (4/12)
            </div>
        </div>
        <div class="col-4">
            <div class="demo-box" style="background-color: #fff9c4; border-color: #ffc107;">
                <strong>col-4</strong><br>One-third (4/12)
            </div>
        </div>
        <div class="col-4">
            <div class="demo-box" style="background-color: #fff9c4; border-color: #ffc107;">
                <strong>col-4</strong><br>One-third (4/12)
            </div>
        </div>
    </div>

    <!-- Four equal columns (3 + 3 + 3 + 3 = 12) -->
    <div class="row mb-3">
        <div class="col-3">
            <div class="demo-box" style="background-color: #f8bbd0; border-color: #e91e63;">
                <strong>col-3</strong><br>Quarter (3/12)
            </div>
        </div>
        <div class="col-3">
            <div class="demo-box" style="background-color: #f8bbd0; border-color: #e91e63;">
                <strong>col-3</strong><br>Quarter (3/12)
            </div>
        </div>
        <div class="col-3">
            <div class="demo-box" style="background-color: #f8bbd0; border-color: #e91e63;">
                <strong>col-3</strong><br>Quarter (3/12)
            </div>
        </div>
        <div class="col-3">
            <div class="demo-box" style="background-color: #f8bbd0; border-color: #e91e63;">
                <strong>col-3</strong><br>Quarter (3/12)
            </div>
        </div>
    </div>

    <!-- Mixed columns (8 + 4 = 12) -->
    <div class="row mb-3">
        <div class="col-8">
            <div class="demo-box" style="background-color: #d1c4e9; border-color: #673ab7;">
                <strong>col-8</strong><br>Two-thirds Width (8/12)
            </div>
        </div>
        <div class="col-4">
            <div class="demo-box" style="background-color: #d1c4e9; border-color: #673ab7;">
                <strong>col-4</strong><br>One-third Width (4/12)
            </div>
        </div>
    </div>
</div>
"""

HTML(bootstrap_grid)
````

### 🔍 Understanding Bootstrap Grid:

- **12-column system** - Total columns always add up to 12
- **`container`** - Centers content and adds padding
- **`row`** - Creates a horizontal row
- **`col-X`** - Creates a column that spans X columns (out of 12)

### Math:
- `col-6` + `col-6` = 12 ✅
- `col-4` + `col-4` + `col-4` = 12 ✅
- `col-8` + `col-4` = 12 ✅
- `col-3` × 4 = 12 ✅

# Part 6: Responsive Design

## What is Responsive Design?

Responsive design means your layout adapts to different screen sizes (mobile, tablet, desktop).

### Bootstrap Breakpoints:

| Class | Screen Size | Example |
|-------|------------|--------|
| `col-` | Extra small (<576px) | Mobile phones |
| `col-sm-` | Small (≥576px) | Mobile landscape |
| `col-md-` | Medium (≥768px) | Tablets |
| `col-lg-` | Large (≥992px) | Laptops |
| `col-xl-` | Extra large (≥1200px) | Desktops |

### Example:
```html
<div class="col-12 col-md-6 col-lg-4">
    <!-- Mobile: Full width (12/12)
         Tablet: Half width (6/12)
         Desktop: One-third width (4/12) -->
</div>
```

# Part 7: Connection to Dash

## How This Relates to Dash

In Dash, you'll use similar concepts but with Python syntax:

### HTML in Dash:
```python
# HTML
<div class="container">
    <h1>Title</h1>
    <p>Paragraph</p>
</div>

# Dash equivalent
html.Div([
    html.H1('Title'),
    html.P('Paragraph')
], className='container')
```

### Bootstrap in Dash:
```python
# HTML
<div class="row">
    <div class="col-6">Left</div>
    <div class="col-6">Right</div>
</div>

# Dash Bootstrap Components
dbc.Row([
    dbc.Col('Left', md=6),
    dbc.Col('Right', md=6)
])
```

````python
# Create sample sales data
np.random.seed(42)
dates = pd.date_range('2023-01-01', '2024-09-30', freq='D')
products = ['Laptop', 'Monitor', 'Keyboard', 'Mouse', 'Headphones']
regions = ['North', 'South', 'East', 'West']
categories = ['Electronics', 'Accessories']

data = []
for _ in range(500):
    product = np.random.choice(products)
    data.append({
        'Date': np.random.choice(dates),
        'Product': product,
        'Region': np.random.choice(regions),
        'Category': 'Electronics' if product in ['Laptop', 'Monitor'] else 'Accessories',
        'Sales': np.random.randint(100, 5000),
        'Quantity': np.random.randint(1, 20),
        'Profit': np.random.randint(20, 1500)
    })

df = pd.DataFrame(data)
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.to_period('M').astype(str)
df['Profit_Margin'] = (df['Profit'] / df['Sales'] * 100).round(2)

print(f"Dataset shape: {df.shape}")
df.head()
````

````python
# Example 1: Simple Dashboard

# Initialize the Dash app
app1 = Dash(__name__)

# Define the layout
app1.layout = html.Div([
    # Header
    html.H1('Sales Dashboard',
            style={'textAlign': 'center', 'color': '#2c3e50', 'marginBottom': 30}),

    # Dropdown filter
    html.Div([
        html.Label('Select Product:', style={'fontSize': 18, 'fontWeight': 'bold'}),
        dcc.Dropdown(
            id='product-dropdown',
            options=[{'label': product, 'value': product} for product in df['Product'].unique()],
            value='Laptop',
            clearable=False,
            style={'width': '300px'}
        )
    ], style={'marginBottom': 30, 'marginLeft': 50}),

    # Graph
    dcc.Graph(id='sales-by-region')

], style={'padding': '20px', 'fontFamily': 'Arial, sans-serif'})

# Define callback
@app1.callback(
    Output('sales-by-region', 'figure'),
    Input('product-dropdown', 'value')
)
def update_graph(selected_product):
    # Filter data
    filtered_df = df[df['Product'] == selected_product]

    # Aggregate by region
    sales_by_region = filtered_df.groupby('Region')['Sales'].sum().reset_index()

    # Create bar chart
    fig = px.bar(
        sales_by_region,
        x='Region',
        y='Sales',
        title=f'Total Sales by Region - {selected_product}',
        color='Sales',
        color_continuous_scale='Blues'
    )

    fig.update_layout(
        title_font_size=20,
        xaxis_title='Region',
        yaxis_title='Total Sales ($)',
        showlegend=False,
        height=500
    )

    return fig

# Run the app
# Uncomment the line below to run
# app1.run_server(debug=True, port=8050)
````
